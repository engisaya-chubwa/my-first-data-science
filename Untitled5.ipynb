```python
import pandas as pd
import matplotlib.pyplot as plt

# Tengeneza data ya mfano ya duka
data = {
    'Mwezi': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Mauzo': [1500000, 1800000, 1200000, 2500000, 3000000, 2800000],
    'Matumizi': [800000, 900000, 700000, 1100000, 1300000, 1200000]
}
df = pd.DataFrame(data)
df['Faida'] = df['Mauzo'] - df['Matumizi']

print("TAARIFA YA DUKA:")
print(df)

# Chora Graph
plt.figure(figsize=(8,4))
plt.plot(df['Mwezi'], df['Mauzo'], marker='o', label='Mauzo')
plt.plot(df['Mwezi'], df['Faida'], marker='s', label='Faida')
plt.title('Mauzo na Faida ya Duka - 2024')
plt.legend()
plt.grid()
plt.show()

print(f"\nMwezi bora ni: {df.loc[df['Faida'].idxmax(), 'Mwezi']} kwa faida ya {df['Faida'].max()}")
```
